```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
```

```python
# 1. Load Raw Datasets
# ---------------------------------------------------------
train_df = pd.read_csv('./data.csv')
test_df = pd.read_csv('./test.csv')

print(f"Train shape: {train_df.shape}, Test shape: {test_df.shape}")
```

```python
# 2. Combine Datasets for Consistent Cleaning & Encoding
# ---------------------------------------------------------
train_df['is_train'] = 1
test_df['is_train'] = 0
df_combined = pd.concat([train_df, test_df], ignore_index=True)
```

```python
# 3. Clean 'Beds' Column
# ---------------------------------------------------------
df_combined['Beds'] = df_combined['Beds'].astype(str).str.lower().str.strip()
df_combined['Beds'] = df_combined['Beds'].replace({
    'studio': '1', 
    'studio+ maid': '1',
    'none': '1',
    'nan': '1',
    '': '1'
})
df_combined['Beds'] = pd.to_numeric(df_combined['Beds'], errors='coerce').fillna(1).astype(int)
```

```python
# 4. Encode Categorical Columns Safely
# ---------------------------------------------------------
categorical_cols = ['Area', 'Governorate', 'Property_type', 'Baths', 'Beds', 'Title', 'Agency', 'Agent_name', 'Availability_date', 'Amenities']

for col in categorical_cols:
    if col in df_combined.columns:
        df_combined[col] = df_combined[col].astype(str).astype('category').cat.codes
```

```python
# 5. Separate Back into Train and Test Sets
# ---------------------------------------------------------
train_clean = df_combined[df_combined['is_train'] == 1].copy()
test_clean = df_combined[df_combined['is_train'] == 0].copy()

drop_cols = ['rent', 'Property_id', 'is_train']

X = train_clean.drop(columns=[c for c in drop_cols if c in train_clean.columns]).copy()
y = pd.to_numeric(train_clean['rent'], errors='coerce')

X_test_final = test_clean.drop(columns=[c for c in drop_cols if c in test_clean.columns]).copy()
```

```python
# 6. Clean Features and Drop Completely Empty Columns
# ---------------------------------------------------------
if 'rent' in X.columns:
    X = X.drop(columns=['rent'])
if 'Rent' in X.columns:
    X = X.drop(columns=['Rent'])

X = X.apply(pd.to_numeric, errors='coerce')
X_test_final = X_test_final.apply(pd.to_numeric, errors='coerce')

X = X.dropna(axis=1, how='all')
X_test_final = X_test_final[X.columns]
```

```python
# 7. Impute Missing Values with Median Directly
# ---------------------------------------------------------
imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(X)
X_test_imputed = imputer.transform(X_test_final)
```

```python
# 8. Clean Target and Align Rows Safely
# ---------------------------------------------------------
df_clean_final = pd.DataFrame(X_imputed)
df_clean_final['target'] = y.values
df_clean_final = df_clean_final.replace([np.inf, -np.inf], np.nan)
df_clean_final = df_clean_final.dropna(subset=['target'])

X_imputed = df_clean_final.drop(columns=['target']).to_numpy()
y = df_clean_final['target'].reset_index(drop=True)
```
