##### 1.❓Are there better or worse days/times in terms of sales? Plot a chart to show the result

```python
# Extracting dates from transaction date

coffee_df['transaction_date'] = pd.to_datetime(coffee_df['transaction_date'])
coffee_df['transaction_day'] = coffee_df['transaction_date'].dt.day_name()
```

```python
# Creating time categories function to return Morning, Noon, Evenging, Night based on transaction_time

def get_time_category(time_val):
    time_str = str(time_val)
    hour = int(time_str.split(':')[0])
    
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Noon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'


# use apply to apply the time category into a new column called time_category

coffee_df['time_category'] = coffee_df['transaction_time'].apply(get_time_category)
```

> 👉 **Write Your Findings Here**: 
-- -

```python
# Are there better or worse days/times in terms of sales?
sales_analysis = coffee_df.groupby(['transaction_day', 'time_category'])['line_item_amount'].sum().unstack()
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
time_order = ['Morning', 'Noon', 'Evening', 'Night']
sales_analysis = sales_analysis.reindex(index=days_order, columns=time_order)

sales_analysis.plot(kind='bar', figsize=(10, 6), edgecolor='black')
plt.title('Sales Revenue by Day and Time Category')
plt.xlabel('Day of the Week')
plt.ylabel('Total Revenue ($)')
plt.xticks(rotation=45)
plt.legend(title='Time of Day')
plt.tight_layout()
plt.show()

# Write your findings:

#Best Time: Morning generates the highest revenue by far across all days, followed by Noon.
#Worst Time: Night has almost zero sales, followed by Evening.
#Best Days: Sales peak slightly on Monday and Wednesday mornings, and reach their lowest points on Sunday.
#Conclusion: The coffee shop is heavily driven by morning commuters and daytime traffic.
```

```python
##### 3.❓Do customers who have been customers longer tend to spend more? Plot a chart to show the result (Hint : You may need to use date.today() to get todays date)
current_date = pd.Timestamp.now()
coffee_df['customer_since'] = pd.to_datetime(coffee_df['customer_since'])
coffee_df['customer_tenure_years'] = (current_date - coffee_df['customer_since']).dt.days / 365.25

customer_spending = coffee_df.groupby('customer_tenure_years')['line_item_amount'].sum().reset_index()

plt.scatter(customer_spending['customer_tenure_years'], customer_spending['line_item_amount'], color='chocolate', alpha=0.6)
plt.title('Customer Tenure vs Total Spending')
plt.xlabel('Tenure (Years)')
plt.ylabel('Total Spending ($)')
plt.tight_layout()
plt.show()
```
