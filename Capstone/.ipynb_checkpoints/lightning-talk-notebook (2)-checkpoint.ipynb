<a id="topic-1"></a>---# Topic 1 : Predicting Airline Passenger Satisfaction

## Project Brief**Problem Statement** -- Airlines collect Likert-scale ratings (1-5) on ~14 touchpoints of a flight (boarding, seat comfort, wifi, entertainment, etc.) plus trip metadata (class, distance, delays), and label each passenger as **satisfied** or **neutral/dissatisfied**. The question -- *which touchpoints actually drive satisfaction, and can we predict it before the passenger tells us?* -- is a binary classification problem. We treat satisfaction as a Bernoulli random variable $Y\in\{0,1\}$ and want to estimate$$P(Y=1\mid X)=\sigma(\beta_0+\beta_1x_1+\dots+\beta_px_p),\qquad \sigma(z)=\frac{1}{1+e^{-z}}$$the log-odds of satisfaction as a linear combination of the service ratings. Each ordinal rating is itself a discrete random variable on $\{1,\dots,5\}$, so part of the EDA is checking how strongly (Pearson's $r$) each co-moves with the binary target.**Goals**- Rank which service dimensions have the strongest linear association with satisfaction.- Build a classifier for $P(\text{satisfied}\mid X)$ that reliably flags likely-dissatisfied passengers (high recall) so an airline could intervene before the flight ends.- Test whether a linear (logistic) boundary suffices, or whether interaction effects between ratings need a non-linear model.**Approach**1. Clean: handle missing `Arrival_Delay_in_Minutes`, drop index columns.2. Encode categoricals and the binary target.3. EDA: target balance, nulls, correlation of every feature with target, outlier check on delays/distance.4. Model: logistic regression baseline vs. random forest / gradient boosting.5. Evaluate with ROC-AUC, precision/recall, confusion matrix.**Data Source(s)** -- `airline_clnd.csv`, Airline Passenger Satisfaction survey data (Kaggle), one row per passenger.**Data Dictionary**| Column | Type | Description || --- | --- | --- || id | int | unique passenger identifier || Gender | str | Male / Female || Customer_Type | str | Loyal Customer / disloyal Customer || Age | int | passenger age in years || Type_of_Travel | str | Business travel / Personal Travel || Class | str | Eco / Eco Plus / Business || Flight_Distance | int | distance flown, miles || Inflight_wifi_service ... Cleanliness | int (1-5) | 14 Likert-scale service ratings (0 = not applicable) || Departure_Delay_in_Minutes | int | minutes departure was delayed || Arrival_Delay_in_Minutes | float | minutes arrival was delayed (has a few nulls) || satisfaction | str | **target**: `satisfied` or `neutral or dissatisfied` |

## EDA

```python
df = pd.read_csv('airline_clnd.csv')
print(df.shape)
df.head()
```

```python
df.info()
```

#### Q: How many null values does each feature have? What's the plan?

```python
nulls = df.isnull().sum()
nulls[nulls > 0]
```

Only `Arrival_Delay_in_Minutes` has missing values (310 rows, ~0.3% of 103,904). It's continuous, right-skewed, and strongly related to `Departure_Delay_in_Minutes` -- plan: median-impute (robust to skew) or drop these rows; either has negligible effect at this rate.

#### Q: Is the target categorical or continuous? Is it balanced?

```python
df['satisfaction'].value_counts(normalize=True).plot(kind='bar', color=['salmon','steelblue'])
plt.title('Target balance: satisfaction')
plt.ylabel('proportion')
plt.show()
df['satisfaction'].value_counts(normalize=True)
```

Categorical (binary): `satisfied` vs `neutral or dissatisfied`. Fairly balanced -- ~56.7% dissatisfied/neutral vs ~43.3% satisfied -- so accuracy is reasonably meaningful here (unlike Topics 2 and 3), though ROC-AUC/F1 are still preferred.

#### Q: What types of features do you have, and how many of each?

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print(f'{len(num_cols)} numeric columns:', num_cols)
print(f'{len(cat_cols)} categorical columns:', cat_cols)
```

#### Q: Which features are most correlated with the target, and with each other?

```python
tmp = df.copy()
tmp['satisfied'] = (tmp['satisfaction'] == 'satisfied').astype(int)
corr_target = tmp.select_dtypes(include=np.number).corr()['satisfied'].drop('satisfied').sort_values()
plt.figure(figsize=(6,8))
corr_target.plot(kind='barh', color=np.where(corr_target>0,'steelblue','salmon'))
plt.title('Pearson r of each feature with satisfaction')
plt.axvline(0, color='k', linewidth=0.8)
plt.show()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(tmp[num_cols[2:]].corr(), cmap='coolwarm', center=0)
plt.title('Feature-feature correlation matrix')
plt.show()
```

`Online_boarding` ($r\approx0.50$) dominates, followed by `Inflight_entertainment` ($r\approx0.40$) and `Seat_comfort` ($r\approx0.35$). `Gate_location` and the two delay columns are essentially uncorrelated ($|r|<0.06$) -- delays barely move satisfaction on their own, a useful, slightly counter-intuitive talking point.

#### Q: Are there outliers? How would you handle them?

```python
fig, axes = plt.subplots(1, 3, figsize=(14,4))
sns.boxplot(y=df['Departure_Delay_in_Minutes'], ax=axes[0]); axes[0].set_title('Departure Delay')
sns.boxplot(y=df['Arrival_Delay_in_Minutes'], ax=axes[1]); axes[1].set_title('Arrival Delay')
sns.boxplot(y=df['Flight_Distance'], ax=axes[2]); axes[2].set_title('Flight Distance')
plt.tight_layout()
plt.show()
```

Both delay columns are heavily right-skewed with a long tail of very late flights (real events, not errors) -- an IQR cutoff would flag thousands of "outliers" that are just rare-but-real delays, so the plan is to **cap/winsorize** or log-transform rather than drop them. `Flight_Distance` has no problematic outliers.

<a id="topic-2"></a>---# Topic 2 : Detecting Credit Card Fraud

## Project Brief**Problem Statement** -- Given a transaction described by 28 anonymized principal components ($V_1,\dots,V_{28}$), the elapsed `Time`, and the `Amount`, predict whether it is fraudulent (`Class=1`) or legitimate (`Class=0`). This is an **extreme class-imbalance** classification / anomaly-detection problem: fraud is rare, $P(Y=1)$ is tiny, so the task is finding a decision region in a 28-dimensional space where the density of legitimate transactions is near zero. The $V_i$ are the output of **PCA** on the original (undisclosed) transaction fields: each $V_i=\mathbf{a}_i^\top\mathbf{x}$, where $\mathbf{a}_i$ is the $i$-th eigenvector of the covariance matrix, ordered by the variance ($\lambda_i$) it explains. By construction the $V_i$ are mutually orthogonal, i.e. $\text{Cov}(V_i,V_j)\approx0$ for $i\ne j$ -- which we verify below.**Goals**- Quantify the imbalance precisely and choose metrics (precision, recall, AUPRC) that stay meaningful under it.- Rank PCA components by $|r(V_i,\text{Class})|$ to find where the fraud signal lives.- Build a model that maximizes fraud recall at an acceptable false-positive rate -- effectively a likelihood-ratio / distance-from-normal (Mahalanobis-style) anomaly score rather than a naive linear cut.**Approach**1. Confirm there are no missing values.2. Measure the imbalance ratio and visualize on a log scale.3. Rank $V_1..V_{28}$ by absolute correlation with `Class`; check the correlation matrix for the expected near-orthogonality.4. Scale `Time`/`Amount` (the only non-PCA features) and examine `Amount` skew/outliers.5. Address imbalance with resampling (SMOTE / class weights); train logistic regression baseline, then a tree ensemble.6. Evaluate with a precision-recall curve, AUPRC, and a confusion matrix at a chosen threshold.**Data Source(s)** -- `creditcard.csv`, Credit Card Fraud Detection dataset (Kaggle / ULB Machine Learning Group), European cardholders' transactions over two days in September 2013.**Data Dictionary**| Column | Type | Description || --- | --- | --- || Time | int | seconds elapsed since the first transaction in the dataset || V1 ... V28 | float | anonymized principal components of the original transaction features || Amount | float | transaction amount || Class | int | **target**: 1 = fraud, 0 = legitimate |

## EDA

```python
df2 = pd.read_csv('creditcard.csv')
print(df2.shape)
df2.head()
```

```python
df2.info()
```

#### Q: How many null values does each feature have?

```python
df2.isnull().sum().sum()  # -> 0: PCA output is dense by construction, no nulls at all
```

#### Q: Is the target categorical or continuous? Is it balanced?

```python
counts = df2['Class'].value_counts()
props = df2['Class'].value_counts(normalize=True)
print(counts)
print(props)

fig, ax = plt.subplots(1,2, figsize=(11,4))
counts.plot(kind='bar', ax=ax[0], color=['steelblue','salmon'])
ax[0].set_yscale('log')
ax[0].set_title('Class counts (log scale)')
ax[0].set_xticklabels(['legit (0)','fraud (1)'], rotation=0)
props.plot(kind='pie', autopct='%1.2f%%', ax=ax[1], colors=['steelblue','salmon'])
ax[1].set_ylabel('')
ax[1].set_title('Class proportion')
plt.tight_layout()
plt.show()
```

Categorical (binary), and severely **imbalanced**: 284,315 legitimate vs. only 492 fraud, i.e. fraud is **0.173%** of transactions (imbalance ratio ~**578:1**). This single number rules out plain accuracy and drives the whole modeling strategy.

#### Q: What types of features do you have, and how many of each?

```python
num_cols2 = df2.select_dtypes(include=np.number).columns.tolist()
cat_cols2 = df2.select_dtypes(exclude=np.number).columns.tolist()
print(f'{len(num_cols2)} numeric columns (PCA output plus Time/Amount/Class)')
print(f'{len(cat_cols2)} categorical columns')
```

#### Q: Which features are most correlated with the target, and with each other?

```python
corr2 = df2.corr()['Class'].drop('Class').sort_values()
plt.figure(figsize=(6,9))
corr2.plot(kind='barh', color=np.where(corr2>0,'steelblue','salmon'))
plt.axvline(0, color='k', linewidth=0.8)
plt.title('Pearson r of each feature with Class')
plt.show()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(df2[num_cols2].corr(), cmap='coolwarm', center=0)
plt.title('Feature-feature correlation matrix (confirms PCA orthogonality: off-diagonal ~ 0)')
plt.show()
```

`V17`, `V14`, `V12`, `V10` ($r\approx-0.2$ to $-0.33$) and, positively, `V11`, `V4` ($r\approx0.13$-$0.15$) stand out; every other component is near zero. `Amount` and `Time` are almost uncorrelated with `Class` individually -- fraud separates mainly through the PCA components.

#### Q: Are there outliers? How would you handle them?

```python
fig, ax = plt.subplots(1,2, figsize=(11,4))
sns.boxplot(y=df2['Amount'], ax=ax[0]); ax[0].set_title('Amount (raw)')
sns.boxplot(y=np.log1p(df2['Amount']), ax=ax[1]); ax[1].set_title('log1p(Amount)')
plt.tight_layout()
plt.show()
df2['Amount'].describe()
```

`Amount` is extremely right-skewed (median \$22, mean \$88, max over \$25,000) -- large-but-legitimate transactions look like IQR outliers but aren't errors, so the plan is a `log1p` transform rather than removal. The $V_i$ columns are already PCA-standardized, and extreme $V_i$ values are often informative (exactly the fraud cases), so no further treatment is needed there.

<a id="topic-3"></a>---# Topic 3 : Detecting Network Infiltration from Flow Statistics

## Project Brief**Problem Statement** -- Network security tools like CICFlowMeter summarize every network *flow* into ~78 statistical features -- packet/byte counts, inter-arrival times (IAT), flags, rates like bytes/second -- labeled `BENIGN` or an attack type (`Infiltration`). This is binary classification in a **high-dimensional** space: each flow is a point $\mathbf{x}\in\mathbb{R}^{78}$, and we want $f(\mathbf{x})\to\{\text{BENIGN},\text{Infiltration}\}$. Several features are literally ratios, e.g.$$\text{Flow Bytes/s}=\frac{\text{Total Fwd Bytes}+\text{Total Bwd Bytes}}{\text{Flow Duration}}$$so a `Flow Duration` of 0 produces a division by zero -- a direct, checkable explanation for the nulls found below, and a reminder that engineered ratio features need care.**Goals**- Characterize the 78-dimensional space: scale differences, null sources, class balance.- Test whether single-feature linear correlation with the label carries any signal at all, or whether classes only separate through feature *combinations* -- which would argue for mutual information / tree-based importance over a correlation heatmap.- Build a classifier that favors **recall** on `Infiltration` (a missed intrusion is far costlier than a false alarm).**Approach**1. Audit shape, dtypes, and the nulls in `Flow Bytes/s`; trace to zero-duration flows and impute.2. Check class balance for `Label`.3. Compute Pearson correlation of every feature with the encoded label.4. Standardize features given huge scale differences (durations up to $10^6$ vs. binary flag columns).5. Reduce dimensionality (PCA) for visualization; drop zero-variance columns.6. Train a tree-based classifier that captures non-linear interactions, using feature importance / mutual information instead of raw correlation.7. Evaluate with recall, F1, and ROC-AUC, prioritizing recall on `Infiltration`.**Data Source(s)** -- `Network_Traffic.csv`, network flow statistics (CICFlowMeter-style features, CIC-IDS lineage) labeled `BENIGN` or `Infiltration`.**Data Dictionary**| Column | Type | Description || --- | --- | --- || Destination Port | int | destination port of the flow || Flow Duration | int | flow duration, microseconds || Total Fwd/Bwd Packets | int | packet counts in each direction || Fwd/Bwd Packet Length Max/Min/Mean/Std | float | packet size statistics per direction || Flow Bytes/s, Flow Packets/s | float | throughput rates (bytes or packets per flow-second) || Flow IAT Mean/Std/Max/Min | float | inter-arrival time statistics between packets || FIN/SYN/RST/PSH/ACK/URG Flag Count | int | TCP flag counts || Active/Idle Mean/Std/Max/Min | float | active vs. idle time statistics of the flow || ... | ... | (78 numeric flow-statistic columns total) || Label | str | **target**: `BENIGN` or `Infiltration` |

## EDA

```python
df3 = pd.read_csv('Network_Traffic.csv')
df3 = df3.drop(columns=[c for c in df3.columns if c.startswith('Unnamed')])
print(df3.shape)
df3.head()
```

```python
df3.info(verbose=False)
```

#### Q: How many null values does each feature have? What's the plan?

```python
nulls3 = df3.isnull().sum()
nulls3[nulls3 > 0]
```

```python
# confirm the divide-by-zero hypothesis
(df3.loc[df3['Flow Bytes/s'].isnull(), 'Flow Duration'] == 0).mean()
```

Only `Flow Bytes/s` has missing values (36 of 488,115 rows -- negligible), corresponding to flows with `Flow Duration == 0` (0/0 undefined). Plan: impute with 0 rather than drop, since dropping would bias against very short-lived flows -- exactly the kind of behavior an intrusion detector should still see.

#### Q: Is the target categorical or continuous? Is it balanced?

```python
counts3 = df3['Label'].value_counts()
props3 = df3['Label'].value_counts(normalize=True)
print(counts3)
print(props3)
props3.plot(kind='bar', color=['steelblue','salmon'])
plt.title('Target balance: Label')
plt.ylabel('proportion')
plt.show()
```

Categorical (binary), reasonably balanced: ~59.1% `BENIGN` vs. ~40.9% `Infiltration`. Unlike Topic 2, this is not a rare-event problem -- the challenge is dimensionality and non-linearity, not imbalance.

#### Q: What types of features do you have, and how many of each?

```python
num_cols3 = df3.select_dtypes(include=np.number).columns.tolist()
cat_cols3 = df3.select_dtypes(exclude=np.number).columns.tolist()
print(f'{len(num_cols3)} numeric columns, {len(cat_cols3)} categorical column(s): {cat_cols3}')
```

#### Q: Which features are most correlated with the target, and with each other?

```python
tmp3 = df3.copy()
tmp3['infil'] = (tmp3['Label'] == 'Infiltration').astype(int)
corr3 = tmp3[num_cols3].corrwith(tmp3['infil']).abs().sort_values(ascending=False)
corr3.head(15)
```

```python
subset = ['Flow Duration','Total Fwd Packets','Total Backward Packets','Flow Bytes/s','Flow Packets/s',
          'Flow IAT Mean','Fwd Packet Length Mean','Bwd Packet Length Mean','Average Packet Size',
          'Init_Win_bytes_forward','Init_Win_bytes_backward','Destination Port']
plt.figure(figsize=(9,7))
sns.heatmap(df3[subset].corr(), cmap='coolwarm', center=0, annot=True, fmt='.2f')
plt.title('Feature-feature correlation (representative subset of 78)')
plt.tight_layout()
plt.show()
```

Even the *strongest* single-feature correlation with `Infiltration` is only $|r|\approx0.0025$ -- essentially zero. No individual flow statistic linearly separates the classes. This is the key finding for this topic: it argues against correlation-driven feature selection or a linear model, and toward non-linear, multivariate methods (tree ensembles, mutual information, PCA-based clustering) that can capture interaction effects a simple $r$ can't see.

#### Q: Are there outliers? How would you handle them?

```python
fig, axes = plt.subplots(1, 3, figsize=(14,4))
sns.boxplot(y=df3['Flow Duration'], ax=axes[0]); axes[0].set_title('Flow Duration')
sns.boxplot(y=df3['Flow Bytes/s'].dropna(), ax=axes[1]); axes[1].set_title('Flow Bytes/s')
sns.boxplot(y=df3['Total Fwd Packets'], ax=axes[2]); axes[2].set_title('Total Fwd Packets')
plt.tight_layout()
plt.show()
```

Duration, byte rates, and packet counts are all extremely right-skewed with orders-of-magnitude outliers -- but in network traffic these extremes (a long-lived connection, a packet burst) are often exactly the anomalous behavior being detected, not noise. Plan: keep them, use scale-robust preprocessing (`RobustScaler` / log transforms) and tree-based models (invariant to monotonic transforms and outlier magnitude) rather than deleting rows.
