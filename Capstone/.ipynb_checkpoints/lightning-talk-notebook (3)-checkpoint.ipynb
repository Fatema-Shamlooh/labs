# Topic 1 : Predicting Airline Passenger Satisfaction

## Problem StatementAirlines survey passengers on ~14 service touchpoints (boarding, seat comfort, wifi, entertainment, cleanliness, etc.) using a 1-5 rating scale, alongside trip details (class, distance, delays), then record whether the passenger was **satisfied** or **neutral/dissatisfied**. The goal is to predict that satisfaction label from the ratings and trip details, and to understand which touchpoints matter most.## Goals- Find out which service ratings are most strongly associated with satisfaction.- Build a classifier that predicts satisfaction from the ratings and trip details.- Check whether delays actually move satisfaction as much as service quality does.## ApproachClean the small number of missing delay values, encode the categorical columns, run EDA on the target and features (balance, nulls, correlation, outliers), then train and evaluate a classifier (logistic regression baseline, then a tree-based model).

## Data InfoData Source : `airline_clnd.csv` -- Airline Passenger Satisfaction survey data (Kaggle), one row per passenger.

## Data Dictionary| Column    | Type    | Description || --------  | ------- | -------     || id        | int     | unique passenger identifier || Gender    | str     | Male / Female || Customer_Type | str | Loyal Customer / disloyal Customer || Age       | int     | passenger age in years || Type_of_Travel | str | Business travel / Personal Travel || Class     | str     | Eco / Eco Plus / Business || Flight_Distance | int | distance flown, miles || Inflight_wifi_service ... Cleanliness | int (1-5) | 14 service ratings || Departure_Delay_in_Minutes | int | minutes departure was delayed || Arrival_Delay_in_Minutes | float | minutes arrival was delayed || satisfaction | str | target: satisfied or neutral or dissatisfied |

```python
# start you EDA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
```

```python
df = pd.read_csv('airline_clnd.csv')
df.head()
```

```python
# What is your target column and what does it represent?
# Is your target categorical or continuous? is it balanced (Plot the distribution)?
df['satisfaction'].value_counts(normalize=True).plot(kind='bar', color=['salmon','steelblue'])
plt.title('Target balance: satisfaction')
plt.ylabel('proportion')
plt.show()
df['satisfaction'].value_counts(normalize=True)
```

The target column is `satisfaction`, representing whether the passenger was satisfied with the flight. It's categorical (binary): `satisfied` vs `neutral or dissatisfied`. It's fairly balanced -- about 56.7% dissatisfied/neutral vs 43.3% satisfied.

```python
# What types of features do you have (categorical / numerical)? how many of each ?
num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print(f'{len(num_cols)} numeric columns:', num_cols)
print(f'{len(cat_cols)} categorical columns:', cat_cols)
```

There are 19 numeric columns (the ratings, age, distance, delays, id) and 5 categorical columns (Gender, Customer_Type, Type_of_Travel, Class, satisfaction).

```python
# How many null values does each feature have? what's your plan for each?
nulls = df.isnull().sum()
nulls[nulls > 0]
```

Only `Arrival_Delay_in_Minutes` has missing values (310 of 103,904 rows, ~0.3%). Plan: impute with the median, since it's right-skewed and closely tied to `Departure_Delay_in_Minutes`.

```python
# Which features are most correlated with your target, and with each other?
tmp = df.copy()
tmp['satisfied'] = (tmp['satisfaction'] == 'satisfied').astype(int)
corr_target = tmp.select_dtypes(include=np.number).corr()['satisfied'].drop('satisfied').sort_values()
plt.figure(figsize=(6,8))
corr_target.plot(kind='barh', color=np.where(corr_target>0,'steelblue','salmon'))
plt.title('Correlation of each feature with satisfaction')
plt.axvline(0, color='k', linewidth=0.8)
plt.show()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(tmp[num_cols[2:]].corr(), cmap='coolwarm', center=0)
plt.title('Feature-feature correlation matrix')
plt.show()
```

`Online_boarding` (r ~ 0.50) is most correlated with satisfaction, followed by `Inflight_entertainment` (r ~ 0.40) and `Seat_comfort` (r ~ 0.35). `Gate_location` and the delay columns are essentially uncorrelated (|r| < 0.06).

```python
# Are there outliers? how would you handle them?
fig, axes = plt.subplots(1, 3, figsize=(14,4))
sns.boxplot(y=df['Departure_Delay_in_Minutes'], ax=axes[0]); axes[0].set_title('Departure Delay')
sns.boxplot(y=df['Arrival_Delay_in_Minutes'], ax=axes[1]); axes[1].set_title('Arrival Delay')
sns.boxplot(y=df['Flight_Distance'], ax=axes[2]); axes[2].set_title('Flight Distance')
plt.tight_layout()
plt.show()
```

Both delay columns have a long tail of large, real delays rather than data errors, so instead of dropping them I'd cap/winsorize or log-transform before modeling. `Flight_Distance` has no problematic outliers.

# Topic 2 : Detecting Credit Card Fraud

## Problem StatementEach transaction is described by 28 anonymized principal components (V1-V28), the elapsed `Time`, and the `Amount`. The goal is to predict whether a transaction is fraudulent (`Class = 1`) or legitimate (`Class = 0`). Fraud is very rare, so this is an extreme class-imbalance problem.## Goals- Quantify how imbalanced the classes are and pick metrics that stay meaningful under that imbalance.- Find which PCA components carry the most fraud signal.- Build a model that catches as much fraud as possible (high recall) without too many false alarms.## ApproachCheck for nulls, measure the imbalance ratio, rank features by correlation with `Class`, scale `Time`/`Amount`, address the imbalance with resampling or class weights, then train and evaluate a classifier using precision-recall/AUPRC rather than accuracy.

## Data InfoData Source : `creditcard.csv` -- Credit Card Fraud Detection dataset (Kaggle / ULB Machine Learning Group), European cardholders' transactions over two days in September 2013.

## Data Dictionary| Column    | Type    | Description || --------  | ------- | -------     || Time      | int     | seconds elapsed since the first transaction || V1 ... V28 | float  | anonymized PCA components of the original transaction features || Amount    | float   | transaction amount || Class     | int     | target: 1 = fraud, 0 = legitimate |

```python
df2 = pd.read_csv('creditcard.csv')
df2.head()
```

```python
# What is your target column and what does it represent?
# Is your target categorical or continuous? is it balanced (Plot the distribution)?
counts = df2['Class'].value_counts()
props = df2['Class'].value_counts(normalize=True)

fig, ax = plt.subplots(1,2, figsize=(11,4))
counts.plot(kind='bar', ax=ax[0], color=['steelblue','salmon'])
ax[0].set_yscale('log')
ax[0].set_title('Class counts (log scale)')
ax[0].set_xticklabels(['legit (0)','fraud (1)'], rotation=0)
props.plot(kind='pie', autopct='%1.2f%%', ax=ax[1], colors=['steelblue','salmon'])
ax[1].set_ylabel('')
ax[1].set_title('Class proportion')
plt.tight_layout()
plt.show()
```

The target column is `Class`, representing whether the transaction was fraud. It's categorical (binary) and severely imbalanced: 284,315 legitimate vs only 492 fraud, i.e. fraud is 0.173% of transactions (about 578:1).

```python
# What types of features do you have (categorical / numerical)? how many of each ?
num_cols2 = df2.select_dtypes(include=np.number).columns.tolist()
cat_cols2 = df2.select_dtypes(exclude=np.number).columns.tolist()
print(f'{len(num_cols2)} numeric columns')
print(f'{len(cat_cols2)} categorical columns')
```

All 31 columns are numeric (Time, V1-V28, Amount, Class); there are no categorical columns.

```python
# How many null values does each feature have? what's your plan for each?
df2.isnull().sum().sum()
```

There are zero null values anywhere in the dataset, so no imputation is needed.

```python
# Which features are most correlated with your target, and with each other?
corr2 = df2.corr()['Class'].drop('Class').sort_values()
plt.figure(figsize=(6,9))
corr2.plot(kind='barh', color=np.where(corr2>0,'steelblue','salmon'))
plt.axvline(0, color='k', linewidth=0.8)
plt.title('Correlation of each feature with Class')
plt.show()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(df2[num_cols2].corr(), cmap='coolwarm', center=0)
plt.title('Feature-feature correlation matrix')
plt.show()
```

`V17`, `V14`, `V12`, `V10` (r ~ -0.2 to -0.33) and `V11`, `V4` (r ~ 0.13-0.15) are most correlated with `Class`; the rest are near zero. The V-features are nearly uncorrelated with each other, which makes sense since they're PCA components.

```python
# Are there outliers? how would you handle them?
fig, ax = plt.subplots(1,2, figsize=(11,4))
sns.boxplot(y=df2['Amount'], ax=ax[0]); ax[0].set_title('Amount (raw)')
sns.boxplot(y=np.log1p(df2['Amount']), ax=ax[1]); ax[1].set_title('log1p(Amount)')
plt.tight_layout()
plt.show()
```

`Amount` is heavily right-skewed (median $22, max over $25,000). These are real large transactions, not errors, so I'd log-transform `Amount` rather than remove them.

# Topic 3 : Detecting Network Infiltration from Flow Statistics

## Problem StatementEvery network flow is summarized into ~78 statistical features (packet/byte counts, inter-arrival times, TCP flags, byte/packet rates) and labeled `BENIGN` or `Infiltration`. The goal is to classify each flow correctly, prioritizing catching real infiltration attempts.## Goals- Understand the scale and structure of the 78 features and where the nulls come from.- Check whether any single feature separates the classes on its own.- Build a classifier that favors recall on `Infiltration`, since a missed intrusion is costly.## ApproachAudit shape, dtypes, and nulls; check class balance; look at correlation of every feature with the label; standardize features given the huge scale differences; then train and evaluate a tree-based classifier that can pick up non-linear patterns.

## Data InfoData Source : `Network_Traffic.csv` -- network flow statistics (CICFlowMeter-style features) labeled BENIGN or Infiltration.

## Data Dictionary| Column    | Type    | Description || --------  | ------- | -------     || Destination Port | int | destination port of the flow || Flow Duration | int | flow duration, microseconds || Total Fwd/Bwd Packets | int | packet counts in each direction || Flow Bytes/s, Flow Packets/s | float | throughput rates || Flow IAT Mean/Std/Max/Min | float | inter-arrival time statistics || FIN/SYN/RST/PSH/ACK/URG Flag Count | int | TCP flag counts || ... | ... | (78 numeric flow-statistic columns total) || Label | str | target: BENIGN or Infiltration |

```python
df3 = pd.read_csv('Network_Traffic.csv')
df3 = df3.drop(columns=[c for c in df3.columns if c.startswith('Unnamed')])
df3.head()
```

```python
# What is your target column and what does it represent?
# Is your target categorical or continuous? is it balanced (Plot the distribution)?
counts3 = df3['Label'].value_counts()
props3 = df3['Label'].value_counts(normalize=True)
props3.plot(kind='bar', color=['steelblue','salmon'])
plt.title('Target balance: Label')
plt.ylabel('proportion')
plt.show()
props3
```

The target column is `Label`, representing whether the flow was benign or an infiltration attack. It's categorical (binary) and reasonably balanced -- about 59.1% BENIGN vs 40.9% Infiltration.

```python
# What types of features do you have (categorical / numerical)? how many of each ?
num_cols3 = df3.select_dtypes(include=np.number).columns.tolist()
cat_cols3 = df3.select_dtypes(exclude=np.number).columns.tolist()
print(f'{len(num_cols3)} numeric columns, {len(cat_cols3)} categorical column(s): {cat_cols3}')
```

There are 79 numeric columns and 1 categorical column (`Label`, the target itself).

```python
# How many null values does each feature have? what's your plan for each?
nulls3 = df3.isnull().sum()
nulls3[nulls3 > 0]
```

Only `Flow Bytes/s` has missing values (36 of 488,115 rows), coming from flows with zero duration (0 bytes / 0 seconds is undefined). Plan: impute with 0 rather than drop those rows.

```python
# Which features are most correlated with your target, and with each other?
tmp3 = df3.copy()
tmp3['infil'] = (tmp3['Label'] == 'Infiltration').astype(int)
corr3 = tmp3[num_cols3].corrwith(tmp3['infil']).abs().sort_values(ascending=False)
corr3.head(10)
```

```python
subset = ['Flow Duration','Total Fwd Packets','Total Backward Packets','Flow Bytes/s','Flow Packets/s',
          'Flow IAT Mean','Fwd Packet Length Mean','Bwd Packet Length Mean','Average Packet Size',
          'Init_Win_bytes_forward','Init_Win_bytes_backward','Destination Port']
plt.figure(figsize=(9,7))
sns.heatmap(df3[subset].corr(), cmap='coolwarm', center=0, annot=True, fmt='.2f')
plt.title('Feature-feature correlation (subset of the 78 features)')
plt.tight_layout()
plt.show()
```

Even the strongest single feature only reaches |r| ~ 0.0025 with `Infiltration` -- essentially no linear relationship. No individual flow statistic separates the classes on its own, so this needs a non-linear, multivariate model rather than correlation-based feature selection.

```python
# Are there outliers? how would you handle them?
fig, axes = plt.subplots(1, 3, figsize=(14,4))
sns.boxplot(y=df3['Flow Duration'], ax=axes[0]); axes[0].set_title('Flow Duration')
sns.boxplot(y=df3['Flow Bytes/s'].dropna(), ax=axes[1]); axes[1].set_title('Flow Bytes/s')
sns.boxplot(y=df3['Total Fwd Packets'], ax=axes[2]); axes[2].set_title('Total Fwd Packets')
plt.tight_layout()
plt.show()
```

Duration, byte rates, and packet counts are all extremely right-skewed with large outliers -- but in network traffic these extremes are often exactly the suspicious behavior being detected. Plan: keep them, but use robust scaling / log transforms and tree-based models rather than deleting rows.
